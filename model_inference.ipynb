```python
import dagshub
import mlflow
import mlflow.sklearn
dagshub.init(repo_owner='kgord23', repo_name='ML_hw_02_IEEE-CIS', mlflow=True)
```

```python
model=mlflow.sklearn.load_model("models:/BestFraudDetectionModel/2")
```

```python
test_transaction=pd.read_csv('/kaggle/input/competitions/ieee-fraud-detection/test_transaction.csv')
test_identity=pd.read_csv('/kaggle/input/competitions/ieee-fraud-detection/test_identity.csv')
```

```python
test=test_transaction.merge(test_identity, on='TransactionID', how='left')
```

```python
predictions=model.predict_proba(test)[:, 1]
```

```python
submission = pd.DataFrame({
    'TransactionID': test['TransactionID'],
    'isFraud': predictions
})
submission.to_csv('submission.csv', index=False)
print(submission.head())
```
